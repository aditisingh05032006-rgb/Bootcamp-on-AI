```python
import numpy as np

arr = np.random.randint(1, 100, size=(5, 5))
print(arr)

row_sums = arr.sum(axis=1)
print(row_sums)

col_sums = arr.sum(axis=0)
print(col_sums)

row_avg = arr.mean(axis=1)
print(row_avg)

col_avg = arr.mean(axis=0)
print(col_avg)
```

```python
import pandas as pd
import numpy as np

data = {
    'name': ['  Aarav  ', 'PRIYA', 'rohan123', 'Sneha!!', 'Karan_45', np.nan],
    'phone': ['+91-9876543210', '9876543211', '98765-43212', '(987) 654-3213', 'abc123', '9876543215'],
    'email': ['aarav@gmail.com', 'PRIYA@YAHOO.com', 'rohan@@mail.com', 'sneha@ com', 'karan@gmail.com', None],
    'salary': ['₹45,000', '$62000', '38,000', '71000.00', 'NA', '48,000']
}

df = pd.DataFrame(data)
print(df)

df['name'] = df['name'].str.strip()
df['name'] = df['name'].str.replace(r'[^a-zA-Z\s]', '', regex=True)
df['name'] = df['name'].str.title()

df['phone'] = df['phone'].str.replace(r'\D', '', regex=True)
df['phone'] = df['phone'].str[-10:]

df['email'] = df['email'].str.lower()
valid_email = df['email'].str.match(r'^[\w\.-]+@[\w\.-]+\.\w+$')
df['valid_email'] = valid_email

df['salary'] = df['salary'].str.replace(r'[^\d.]', '', regex=True)
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')

df['name'] = df['name'].replace('', np.nan)

df = df.dropna(subset=['name'])

print(df)
```

```python
import matplotlib.pyplot as plt

height = ['140-149', '150-159', '160-169', '170-179', '180-189', '190-199']

people = [5, 12, 20, 25, 15, 5]

plt.barh(height, people)

plt.xlabel("Number of People")
plt.ylabel("Height (cm)")
plt.title("Height-wise People Graph")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2 * np.pi, 1000)

y = np.sin(x)

plt.plot(x, y)

plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)

plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Sine Wave")

plt.grid(True)
plt.show()
```
